```
#Labelling Stock Price 'Close'
#Labelling the data
def create_labels(df, col_name):

        #Label code : BUY => 1, SELL => 0, HOLD => 2

        window_size = 11
        row_counter = 0
        total_rows = len(df)
        labels = [2 for i in range(total_rows)]

        window_begin_idx=0; window_end_idx=0; window_middle_idx=0; min_idx=0; max_idx=0; number=0.0; mins=10000.0; maxs=0.0


        while row_counter < total_rows:
            if row_counter > window_size-1:
                window_begin_idx = row_counter - window_size
                window_end_idx = window_begin_idx + window_size - 1
                window_middle_idx = (window_begin_idx + window_end_idx) // 2


                for i in range(window_begin_idx, window_end_idx):
                    price = df.iloc[i][col_name]

                    if price < mins:
                      mins = price
                      min_idx = i

                    if price > maxs:
                      maxs = price
                      max_idx = i

                if max_idx == window_middle_idx:
                  labels[window_middle_idx] = 0

                elif min_idx == window_middle_idx:
                  labels[window_middle_idx] = 1

                else:
                  labels[window_middle_idx] = 2

                mins = 10000.0
                maxs = 0.0

            row_counter+=1

        return labels
```

# CNN

Hyperparameter yang akan diuji adalah :
1. Kernel Size for Convolution -> 3,5,7

2. Number of Kernels (in two Convolution Layers) -> 6:13, 19:38, 32:64, 64:128

3. Number of Parameter in Hidden Layer -> 100, 200

4. Learning Algorithm -> Adam, AdaGrad, AdaDelta, Momentum SGD

5. Activation -> Softmax, Linear, Sigmoid

Pada bawah ini, model CNN 1D hanya merupakan opsional, karena ingin melihat seberapa akurat model CNN 1D dapat memprediksi harga saham, namun karena fluktuasi yang cukup tinggi, model tidak dapat memprediksi saham dengan cukup baik

```
model = Sequential()
model.add(Convolution1D(112, 15, input_shape= (225,1), padding='same',  activation='relu'))
model.add(MaxPooling1D(pool_size=4))
model.add(Convolution1D(225, 15, padding='same'))
model.add(MaxPooling1D(pool_size=2))

# flatten output of conv
model.add(Flatten())

#Hidden Layer
model.add(Dense(400, activation='relu'))

#Output Layer
model.add(Dense(1))
model.add(Activation('linear'))
model.summary()

model.compile(loss='mean_squared_error', optimizer='rmsprop', metrics=['accuracy'])
model.fit(x_train.reshape(x_train.shape[0], 225, 1), price_train, batch_size=4901, epochs=1000, validation_data = (x_test.reshape(x_test.shape[0], 225, 1), price_test), verbose=2)

```

```
# building a linear stack of layers with the sequential model
model1 = Sequential()
# convolutional layer
model1.add(Conv2D(13, kernel_size=(3,3), strides=(1,1), padding='valid', activation='relu', input_shape=(15,15,1)))
model1.add(MaxPool2D(pool_size=(1,1)))
model1.add(Conv2D(6, kernel_size=(3,3), strides=(1,1), padding='valid'))
model1.add(MaxPool2D(pool_size=(1,1)))

# flatten output of conv
model1.add(Flatten())
# hidden layer
model1.add(Dense(100, activation='relu'))
# output layer
model1.add(Dense(3, activation='softmax'))

# compiling the sequential model
model1.compile(loss='categorical_crossentropy', metrics=['accuracy'], optimizer='adagrad')

# training the model for 1000 epochs
model1.fit(x_train, y_train_One_hot, batch_size=4901, epochs=1000, validation_data=(x_test, y_test_One_hot))
```

# Accuracy

```
Prediction = model1.predict(x_train)
Predictions = []
Index = []
for i in range(len(x_train)):
  Max = Prediction[i].max()
  for j in range(3):
    if Prediction[i][j] == Max:
      Predictions.append(j)
      Index.append(i)
```

```
Test = model1.predict(x_test)
Tests = []
Indexs = []
for i in range(len(x_test)):
  Maxs = Test[i].max()
  for j in range(3):
    if Test[i][j] == Maxs:
      Tests.append(j)
      Indexs.append(i)
```

```
#Duplicate Searching
u = 0

for i in range(len(y_train)):
  for j in range(i+1, len(y_train)):
    if Index[i] == Index[j]:
      Predictions.pop(i+u)
      Index.pop(i+u)
      u = u + 1

uu = 0

for i in range(len(y_test)):
  for j in range(i+1, len(y_test)):
    if Indexs[i] == Indexs[j]:
      Tests.pop(i+uu)
      Indexs.pop(i+uu)
      uu = uu + 1
```

```
CC = Predictions
JumlahHold = 0
JumlahSell = 0
JumlahBuy = 0
for i in range(len(CC)):
  if CC[i] == 2 :
    JumlahHold += 1
  if CC[i] == 1 :
    JumlahBuy += 1
  if CC[i] == 0 :
    JumlahSell += 1

DD = Tests
for i in range(len(DD)):
  if DD[i] == 2 :
    JumlahHold += 1
  if DD[i] == 1 :
    JumlahBuy += 1
  if DD[i] == 0 :
    JumlahSell += 1


print("Jumlah Prediksi Label Saham Hold adalah :",JumlahHold)
print("Jumlah Prediksi Label Saham Buy adalah :",JumlahBuy)
print("Jumlah Prediksi Label Saham Sell adalah :",JumlahSell)
```

```
#Confusion Matrix
HoldHold = []
HoldBuy = []
HoldSell = []
SellHold = []
SellBuy = []
SellSell = []
BuyHold = []
BuyBuy = []
BuySell = []
for i in range(len(Predictions)):
  if y_train[i] == 2:
    if Predictions[i] == 2:
      HoldHold.append(i)
    if Predictions[i] == 1:
      HoldBuy.append(i)
    if Predictions[i] == 0:
      HoldSell.append(i)
  elif y_train[i] == 0:
    if Predictions[i] == 2:
      SellHold.append(i)
    if Predictions[i] == 1:
      SellBuy.append(i)
    if Predictions[i] == 0:
      SellSell.append(i)
  elif y_train[i] == 1:
    if Predictions[i] == 2:
      BuyHold.append(i)
    if Predictions[i] == 1:
      BuyBuy.append(i)
    if Predictions[i] == 0:
      BuySell.append(i)

for i in range(len(Tests)):
  if y_test[i] == 2:
    if Tests[i] == 2:
      HoldHold.append(i)
    if Tests[i] == 1:
      HoldBuy.append(i)
    if Tests[i] == 0:
      HoldSell.append(i)
  elif y_test[i] == 0:
    if Tests[i] == 2:
      SellHold.append(i)
    if Tests[i] == 1:
      SellBuy.append(i)
    if Tests[i] == 0:
      SellSell.append(i)
  elif y_test[i] == 1:
    if Tests[i] == 2:
      BuyHold.append(i)
    if Tests[i] == 1:
      BuyBuy.append(i)
    if Tests[i] == 0:
      BuySell.append(i)
```

```
#Precision, Recall, F1 Score
#a. Hold
PrecisionHold = len(HoldHold)/JumlahHold
RecallHold = len(HoldHold)/hold
F1Hold = 2*((RecallHold*PrecisionHold)/(RecallHold + PrecisionHold))
print("Precision pada Hold:",PrecisionHold)
print("Recall pada Hold:",RecallHold)
print("F1Hold pada Hold:",F1Hold)

#b. Sell
PrecisionSell = len(SellSell)/JumlahSell
RecallSell = len(SellSell)/sell
F1Sell = 2*((RecallSell*PrecisionSell)/(RecallSell + PrecisionSell))
print("Precision pada Sell:",PrecisionSell)
print("Recall pada Sell:",RecallSell)
print("F1Sell pada Sell:",F1Sell)

#c. Buy
PrecisionBuy = len(BuyBuy)/JumlahBuy
RecallBuy = len(BuyBuy)/buy
F1Buy = 2*((RecallBuy*PrecisionBuy)/(RecallBuy + PrecisionBuy))
print("Precision pada Buy:",PrecisionBuy)
print("Recall pada Buy:",RecallBuy)
print("F1Buy pada Buy:",F1Buy)
```
